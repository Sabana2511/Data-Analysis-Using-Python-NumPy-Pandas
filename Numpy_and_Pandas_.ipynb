# **Numpy and Pandas**
## **Numpy Array Operations:**

```python
# import numpy and pandas:

import numpy as np
import pandas as pd

# Create a 1D NumPy Array:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("\n Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# Inspection and Properties

print("\nInspection and Properties:")
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# Array Operations

# Convert Celsius to Fahrenheit

temp_fahrenheit = (temperatures_w1 * 9/5) + 32

print("\nTemperatures in Fahrenheit:")
print(temp_fahrenheit)

# Maximum, Minimum and Mean

max = np.max(temperatures_w1)
min = np.min(temperatures_w1)
mean = np.mean(temperatures_w1)

print("\nMaximum Temperature:", max )
print("Minimum Temperature:", min )
print("Mean Temperature:",mean )
```

```python
# Array Slicing and Indexing

# First three days
first_three_days = temperatures_w1[:3]

# Weekend - last two days
weekend = temperatures_w1[-2:]

# Middle three days
middle_three_days = temperatures_w1[2:5]

print("\nArray Slicing:")
print("First Three Days:", first_three_days)
print("Weekend:", weekend)
print("Middle Three Days:", middle_three_days)
```

```python
# Create a 2D NumPy Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # (Week 1)
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # (Week 2)
])

print("\nTemperatures for Two Weeks:")
print(temperatures)
```

```python
# Inspect and Slice the 2D Array

print("\n2D Array Properties:")
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
# Extract each week
week1 = temperatures[0]
week2 = temperatures[1]

# Weekend temperatures for both weeks
week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]

print("\nWeek 1:", week1)
print("Week 2:", week2)

print("\nWeek 1 Weekend:", week1_weekend)
print("Week 2 Weekend:", week2_weekend)
```

## **Pandas Series**

```python
# Creating Pandas Series

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("\nMarks Series:")
print(marks)
```

```python
# Indexing and Slicing

# Integer index position - 1st rank
first_rank = marks.iloc[0]

# Top 3 ranks using loc
top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

# 3rd ranked student using iloc
third_rank = marks.iloc[2]

# Boolean mask - marks greater than 90
marks_greater_90 = marks[marks > 90]

print("\nIndexing and Slicing:")
print("1st Rank Mark:", first_rank)

print("\nTop 3 Ranks:")
print(top_three)

print("\n3rd Rank Mark:", third_rank)

print("\nMarks Greater Than 90:")
print(marks_greater_90)
```

```python
# Manipulating Series

# Modify Rank1 mark to 100
marks.loc['Rank1'] = 100

# Remove Rank5
marks = marks.drop('Rank5')

# Calculate CGPA
cgpa = marks / 10

print("\nManipulating Series:")
print("Updated Marks:")
print(marks)

print("\nCGPA:")
print(cgpa)
```

## **Pandas DataFrame**

```python
# Creating DataFrame

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [
        200, 150, 300, 450, 200,
        250, 300, 180, 350, 400
    ]
})

print("\nTransactions DataFrame:")
print(transactions)
```

```python
# DATA EXPLORATION

# Display DataFrame
print("\nFull DataFrame:")
print(transactions)
```

```python
# Head
print("\nHead:")
print(transactions.head())
```

```python
# Tail
print("\nTail:")
print(transactions.tail())
```

```python
# Column Names
print("\nColumn Names:")
print(transactions.columns)
```

```python
# Data Types
print("\nData Types:")
print(transactions.dtypes)
```

```python
# Basic Information
print("\nBasic Information:")
transactions.info()
```

```python
# Display ProductCategory and Amount
print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve last 3 columns
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])
```

```python
# Filter Region = North and Amount > 200
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print("\nNorth Region with Amount Greater Than 200:")
print(filtered_transactions)
```

```python
# Value counts for ProductCategory
print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())
```

```python
# Unique values in Region
print("\nUnique Regions:")
print(transactions['Region'].unique())
```

```python
# Group by Region and find mean Amount
region_mean = transactions.groupby('Region')['Amount'].mean()

print("\nMean Amount by Region:")
print(region_mean)
```

```python
# DATAFRAME MANIPULATION

# Modify Amount for TransactionID 102 to 165

transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print("\nAfter modifying TransactionID 102:")
print(transactions)
```

```python
# Add Discount column = 10% of Amount

transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter Adding Discount Column:")
print(transactions)
```

```python
# Remove row with TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

print("\nAfter Removing TransactionID 109:")
print(transactions)
```

```python
# Delete Discount column

transactions = transactions.drop(columns=['Discount'])

print("\nAfter Deleting Discount Column:")
print(transactions)
```
